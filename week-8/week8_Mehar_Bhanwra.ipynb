# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

````python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)
````

````python
import re

# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        # Remove extra spaces
        expression = expression.strip()

        # Allow only numbers, operators, parentheses, and decimal points
        if not re.fullmatch(r"[0-9+\-*/(). ]+", expression):
            return "Invalid mathematical expression."

        result = eval(expression)

        return str(result)

    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract up to 5 unique keywords from text."""
    try:
        # Remove punctuation
        text = re.sub(r"[^\w\s]", "", text)

        words = text.lower().split()

        # Remove common stopwords
        stopwords = {
            "the", "this", "that", "with", "from", "have", "will",
            "into", "about", "there", "their", "which", "would",
            "could", "should", "these", "those", "after", "before",
            "under", "over", "your", "while", "where", "when", "what",
            "extract", "keyword", "keywords", "and", "for", "are",
            "was", "were", "has", "had", "our", "you", "they", "them",
            "then", "than", "also", "only"
        }

        keywords = []

        for word in words:
            if len(word) > 4 and word not in stopwords and word not in keywords:
                keywords.append(word)

        return keywords[:5]

    except Exception:
        return []
````

# BONUS TOOL

````python
# 🛠️ TOOL 3: Word Counter

def word_counter(text: str) -> int:
    """Count the number of words in the given text."""
    try:
        return len(text.split())
    except Exception:
        return 0
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
import re

def agent(query: str):

    logging.info(f"Received query: {query}")

    query_lower = query.lower().strip()

    calculator_words = ["calculate", "calc", "compute", "solve"]
    keyword_words = ["keyword", "keywords", "extract"]
    word_count_words = ["count words", "word count", "count"]

    # ---------------- Calculator ----------------

    if any(word in query_lower for word in calculator_words):

        logging.info("Routing to Calculator Tool")

        expression = query

        for word in calculator_words:
            expression = re.sub(word, "", expression, flags=re.IGNORECASE)

        expression = expression.strip()

        result = calculator(expression)

        logging.info(f"Calculator Result: {result}")

        return {
            "type": "calculator",
            "input": expression,
            "result": result
        }

    # ---------------- Keyword Extractor ----------------

    elif any(word in query_lower for word in keyword_words):

        logging.info("Routing to Keyword Extraction Tool")

        text = query

        phrases = [
            "extract keywords from",
            "extract keyword from",
            "extract keywords",
            "extract keyword",
            "the following sentence:",
            "following sentence:"
        ]

        for phrase in phrases:
            text = re.sub(phrase, "", text, flags=re.IGNORECASE)

        text = text.strip()

        result = extract_keywords(text)

        logging.info(f"Keywords: {result}")

        return {
            "type": "keyword_extractor",
            "input": text,
            "result": result
        }

    # ---------------- Word Counter ----------------

    elif any(word in query_lower for word in word_count_words):

        logging.info("Routing to Word Counter Tool")

        text = query

        for phrase in word_count_words:
            text = re.sub(phrase, "", text, flags=re.IGNORECASE)

        text = text.strip()

        result = word_counter(text)

        logging.info(f"Word Count: {result}")

        return {
            "type": "word_counter",
            "input": text,
            "result": result
        }

    # ---------------- Unknown ----------------

    logging.warning("No matching tool found.")

    return {
        "type": "unknown",
        "result": "Sorry, I can only perform calculations, extract keywords, or count words."
    }
````
